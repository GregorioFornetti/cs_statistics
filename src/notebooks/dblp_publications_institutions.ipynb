```python
from pathlib import Path
from lxml import etree
```

```python
data_path = '../../data/raw/dblp.xml'
```

```python
parser = etree.iterparse(
    data_path,
    events=('end',), 
    huge_tree=True,
    tag=("article","inproceedings",)
)
```

```python
i = 0

for _, element in parser:
    i += 1
    if i == 100:
        break
    # print the xml formated data
    print(etree.tostring(element, pretty_print=True).decode('utf-8'))
    print('-----------------------------------')
```

It seems that the XML does not have the affiliation data. Maybe we can extract it with the RDF file?

```python
data_path = '../../data/raw/dblp.ttl'
```

```python
from pyoxigraph import Store, RdfFormat

# Persistent database stored on disk
store = Store("../../data/raw/dblp_store")
```

```python
# Run this once to import the dataset
store.bulk_load(
    path=data_path,
    format=RdfFormat.TURTLE,
)
```

```python
query = """
SELECT ?type (COUNT(*) AS ?count)
WHERE {
    ?subject a ?type .
}
GROUP BY ?type
ORDER BY DESC(?count)
LIMIT 20
"""

for row in store.query(query):
    print(row["type"].value, row["count"].value)
```
